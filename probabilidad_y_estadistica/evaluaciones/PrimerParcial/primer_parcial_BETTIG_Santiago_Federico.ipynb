#### Resolución Ejercicio 1:

Se tiene 10 monedas, 6 justas y 4 falsas, por  lo que la probabilidad de sacar de la bolsa cada una de ellas será:

$$
P(J) = \frac{6}{10}
$$ 

$$P(F) = \frac{4}{10}$$

Por otro lado, la probabilidad de sacar cara habiendo sacando una u otro tipo de moneda será:

$$P(C | J) = \frac{5}{10}$$

$$P(C | F) = \frac{4}{10}$$

La probabilidad de obtener 13 caras seguidas en cada uno de estos casos será:

$$P(C_{13} | J) = P(C | J)^{13} = 0.0001221$$

$$P(C_{13} | F) = P(C | F)^{13} = 0.0000672$$

Aplicando el teorema de probabilidad total, podemos obtener la probabilidad de obtener 13 caras seguidas contemplando todos los casos:

$$P(C_{13}) = P(C_{13} | J) . P(J) + P(C_{13} | F) . P(F)$$

$$P(C_{13})= 0.0001221 * \frac{6}{10} + 0.0000672 * \frac{4}{10}$$

$$P(C_{13}) = 0.00010014$$


Por último, aplico el Teoréma de Bayes para obtener las probabilidades de que la moneda sea una o la otra, sabiendo que salió cara 13 veces seguidas:

$$\boxed{P(J | C_{13}) = \frac{P(C_{13} | J).P(J)}{P(C_{13})} = 0.732}$$

$$\boxed{P(F | C_{13}) = \frac{P(C_{13} | F).P(F)}{P(C_{13})} = 0.268}$$

a) Se tiene la siguiente función densidad definda por partes
$$
f(x, y) =
\begin{cases}
ky & \text{si } x^2 \leq y \leq 2x \\
0  & \text{otro caso }
\end{cases}
$$

Lo primero que se debe hacer es obtener el rango de integración. El de $y$ viene dado por la consigna pero el de $x$ debe deducirse:

$$
x^2 \leq 2x \\
x^2 - 2x \leq 0 \\
x (x - 2) \leq 0 \\
0 \leq x \leq 2
$$

Sabiendo que la integral de la función densidad, definida dentro de todo el rango es igual a 1, se pude obtener k:

$$
1 = \int_{0}^{2} \int_{x^2}^{2x} ky \, dy \, dx
$$

$$
1 = \int_{0}^{2} k \Big[\frac{y^2}{2}\Big]_{x^2}^{2x} \, dx = k \int_{0}^{2} 2x^2 - \frac{x^4}{2} \, dx
$$

$$
1 = k \Big[\frac{2x^3}{3}-\frac{x^5}{10}\Big]_{0}^{2} = k [\frac{16}{3} - \frac{32}{10}] \\
$$

$$
\boxed{k = \frac{15}{32}}
$$

b) La densidad marginal se obtiene de la siguiente manera:

$$ 
f_{x}(x) = \int_{x^2}^{2x}  \frac{15}{32} y  \, dy = \frac{15}{32} \Big[\frac{y^2}{2}\Big]_{x^2}^{2x}
$$

$$ 
f_{x}(x) = \frac{15}{32} [2x^2 - \frac{x^4}{2}]
$$

$$ 
\boxed{f_{x}(x) = \frac{15}{64} [4x^2 - x^4]}
$$

c) Antes de obtener la densidad marginal de $Y$, debo conocer cuales son los límites de integración. Los mismos pueden deducirse a partir del rango dado:

y \leq 2x  \Rightarrow \frac{y}{2} \leq x\\
y \geq x^2 \Rightarrow \sqrt{y} \geq x
\end{cases}
$$

$$
\frac{y}{2} \leq x \leq \sqrt{y}
$$

Conociendo los límites de integración, se puede obtener la densidad marginal:

$$ 
f_{y}(y) = \int_{\frac{y}{2}}^{\sqrt{y}}  \frac{15}{32} y  \, dx = \frac{15}{32} y \Big[x\Big]_{\frac{y}{2}}^{\sqrt{y}}
$$

$$ 
\boxed{f_{y}(y) = \frac{15}{32}[y^{\frac{3}{2}} - \frac{y^2}{2}]}
$$

d) Para obtener la función densidad condicional $f_{X|Y}(x|y)$ se debe realizar el siguiente cociente:

$$
f_{X|Y}(x|y) = \frac{f(x, y)}{f_{x}(x)}

$$
f_{X|Y}(x|y) = \frac{\frac{15}{32}y}{\frac{15}{32}[y^{\frac{3}{2}} - \frac{y^2}{2}]}

#### Resolución Ejercicio 3
Se tiene:
$$Y = R\cos(2\pi \cdot \Phi)$$
donde:
- $\Phi \sim \text{Uniforme}(0, 1)$, con densidad $f_\Phi(\phi) = 1$ para $0 \leq \phi \leq 1$
- $R \sim \text{Rayleigh}$, con densidad $f_R(r) = re^{-r^2/2}$ para $r > 0$
- $R$ y $\Phi$ son **independientes**

Como $R$ y $\Phi$ son independientes, su densidad conjunta es el producto de sus densidades marginales:
 
$$f_{R,\Phi}(r,\phi) = f_R(r) \cdot f_\Phi(\phi) = re^{-r^2/2} \cdot 1 = re^{-r^2/2}$$
 
para $r > 0$ y $0 \leq \phi \leq 1$.

Necesito obtener la densidad conjunta $f_{X,Y}(x,y)$. Para ello, voy a valerme de la fórmula $X^2 + Y^2 = R^2$, despejar $R(X, Y)$ y $\Phi(X, Y)$ y obtener el Jacobiano:

$$R = \sqrt{X^2 + Y^2} \qquad \Phi = \frac{1}{2\pi}\arctan\left(\frac{Y}{X}\right)$$
 
El Jacobiano es el determinante de la matriz de derivadas parciales:
 
$$J = \begin{vmatrix} \frac{\partial r}{\partial x} & \frac{\partial r}{\partial y} \\ \frac{\partial \phi}{\partial x} & \frac{\partial \phi}{\partial y} \end{vmatrix}$$
 
Calculando cada derivada parcial:
 
$$\frac{\partial r}{\partial x} = \frac{x}{\sqrt{x^2+y^2}} \qquad \frac{\partial r}{\partial y} = \frac{y}{\sqrt{x^2+y^2}}$$
 
$$\frac{\partial \phi}{\partial x} = \frac{-y}{2\pi(x^2+y^2)} \qquad \frac{\partial \phi}{\partial y} = \frac{x}{2\pi(x^2+y^2)}$$
 
Por lo tanto, el Jacobiano resulta:
 
$$|J| = \frac{x}{\sqrt{x^2+y^2}} \cdot \frac{x}{2\pi(x^2+y^2)} - \frac{y}{\sqrt{x^2+y^2}} \cdot \frac{-y}{2\pi(x^2+y^2)}$$
 
$$= \frac{x^2 + y^2}{2\pi(x^2+y^2)\sqrt{x^2+y^2}} = \frac{1}{2\pi\sqrt{x^2+y^2}}$$
 

Habiendo obtenido el Jacobiano, puedo obtener la densidad cconjunta aplicando la siguiente ecuación:
$$
f_{X,Y}(x,y) = f_{R,\Phi}(r,\phi) \cdot |J|
$$
$$
f_{X,Y}(x,y) = \sqrt{x^2+y^2} \cdot e^{-(x^2+y^2)/2} \cdot \frac{1}{2\pi\sqrt{x^2+y^2}}
$$
$$
f_{X,Y}(x,y) = \frac{1}{2\pi} \cdot e^{-(x^2+y^2)/2}
$$

Por último, debo obtener la densidad marginal de $Y$. Para ello, integro en todo el rango de $X$ la función densidad conjunta obtenida en el paso anterior:

$$
f_Y(y) = \int_{-\infty}^{\infty} f_{X,Y}(x,y)  \, dx
$$

$$
f_Y(y) = \int_{-\infty}^{\infty} \frac{1}{2\pi} e^{-(x^2+y^2)/2} \, dx = \frac{e^{-y^2/2}}{2\pi} \int_{-\infty}^{\infty} e^{-x^2/2} \, dx
$$

$$
f_Y(y) = \frac{e^{-y^2/2}}{2\pi} \cdot \sqrt{2\pi} = \frac{1}{\sqrt{2\pi}} e^{-y^2/2}
$$

$$
\boxed{f_Y(y) = \frac{1}{\sqrt{2\pi}} e^{-y^2/2}}
$$

Igualando esta última ecuación a la de una distribución normal genérica obtengo:

$$
f_Y(y) = \frac{1}{\sqrt{2\pi}} e^{\frac{-y^2}{2}} = \frac{1}{\sigma\sqrt{2\pi}} e^{\frac{-(y-\mu)^2}{2\sigma^2}}
$$
- $\sigma = 1$
- $\mu = 0$

Y con esto último queda comprobado que $Y$ tiene una distribución normal

#### Resolución Ejercicio 4_a

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from scipy.stats import gaussian_kde
from scipy.interpolate import interp1d
 
np.random.seed(25)

# Efectos según el año
efecto_anual = {2023: 10, 2024: 15, 2025: 20}

# Efectos según el mes (El índice del diccionario representa el número de mes)
efecto_mensual = {
    1: 10, 2: 15, 3: 20, 4: 20, 5: 25, 6: 25,
    7: 30, 8: 25, 9: 25, 10: 20, 11: 15, 12: 10
}
 
# 0=Lunes, ..., 6=Domingo en pandas/python
efecto_diario = {
    6: 10,  # Domingo
    0: 20,  # Lunes
    1: 30,  # Martes
    2: 35,  # Miércoles
    3: 30,  # Jueves
    4: 20,  # Viernes
    5: 10   # Sábado
}

# Efecto de cada local
efecto_local = {"Don Francisco": 20, "Don Miguel": 10}
 
# La función permite simular el efecto diario de acuerdo
# al proietario que se pase como parámetro
def simular_clientes(propietario):
    # Creo una lista con todos los días desde el 01/01/2023 hasta el 
    # 31/12/2025
    fechas = pd.date_range(start="2023-01-01", end="2025-12-31", freq="D")

    # Creo una lista vacia en donde guardar la cantidad de clientes de
    # acuerdo al día
    registros = []
    
    # Recorro las fechas una por una y voy cargando las contribuciones 
    # de los clientes en la lista "registros"
    for fecha in fechas:
        # El lambda será la sumatoria de todos los efectos antes
        # mencionados
        lam = (
            efecto_anual[fecha.year]
            + efecto_mensual[fecha.month]
            + efecto_diario[fecha.dayofweek]
            + efecto_local[propietario]
        )

        # La contribución del cliente tendrá una distribución poisson
        # cuyo lambda fue calculado anteriormente
        clientes = np.random.poisson(lam)
        registros.append({
            "fecha": fecha,
            "año": fecha.year,
            "mes": fecha.month,
            "dia_semana": fecha.day_name(),
            "lambda": lam,
            "clientes": clientes,
            "propietario": propietario
        })
    
    # Convierto los resultados en un dataframe para facilitar el manejo 
    # de los datos.
    return pd.DataFrame(registros)
 
# Obtengo los datos de los clientes para ambos negocios
df_francisco = simular_clientes("Don Francisco")
df_miguel    = simular_clientes("Don Miguel")

# Imprimo los primeros 5 resultados de cada uno de los dataframes.
print("\nDon Francisco, 5 primeras filas:")
print(df_francisco[["fecha", "dia_semana", "lambda", "clientes"]].head().to_string(index=False))
print("\nDon Miguel, 5 primeras filas:")
print(df_miguel[["fecha", "dia_semana", "lambda", "clientes"]].head().to_string(index=False))
 
# Imprimo estadísticas básicas de cada uno de los negocios.
print("\nEstadísticas Anuales de Don Francisco:")
print(df_francisco.groupby("año")["clientes"].describe().round(2))
print("\nEstadísticas Anuales de Don Miguel")
print(df_miguel.groupby("año")["clientes"].describe().round(2))
```

#### Ejercicio 4b y 4c

```python
# Con esta función voy a imprimir la función de distribución acumulada
# y la densidad de cada uno de los registros
def plot_fda_densidad(df, propietario, ax_fda, ax_den, año, color):
    # Voy a quedarme únicamente los datos de los clientes, ya que las 
    # fechas no me interesan
    datos = df[df["año"] == año]["clientes"].values
    # Ordeno los datos de menor a mayor para que los datos tengan sentido
    # Recordar que la FDA es monotonamente creciente
    datos_sorted = np.sort(datos)
    # Obtengo la longitud del array
    n = len(datos_sorted)
 
    # FDA empírica (escalón)
    # Creo un vector con los valores del eje X
    fda = np.arange(1, n + 1) / n
    # Genero una función escalonada con los valores previamente ordenados
    ax_fda.step(datos_sorted, fda, where="post", color=color, label=str(año))
    # Fijo algunas propiedades del gráfico (Título, labels, etc)
    ax_fda.set_title(f"FDA Empírica — {propietario}")
    ax_fda.set_xlabel("Clientes diarios")
    ax_fda.set_ylabel("F(x)")
    ax_fda.legend()
    ax_fda.grid(True, alpha=0.3)
 
    # Densidad empírica (KDE)
    # Obtengo la función kde utilizando gaussian_kde, utilizo el método
    # "scott" para que la curva sea realtivamente suave
    kde = gaussian_kde(datos, bw_method="scott")
    # Genero los valores que tendrá el eje X, le doy margen para que 
    # se vea bien y no parece que se corta abruptamente. Utilizo 
    # 300 puntos en total para realizar el gráfico, esto le da
    # resolución.
    x = np.linspace(datos.min() - 10, datos.max() + 10, 300)
    # Configuro propiedades del gráfico
    ax_den.plot(x, kde(x), color=color, label=str(año))
    ax_den.set_title(f"Densidad Empírica — {propietario}")
    ax_den.set_xlabel("Clientes diarios")
    ax_den.set_ylabel("f(x)")
    ax_den.legend()
    ax_den.grid(True, alpha=0.3)
 
# Defino los colores que voy a usar para distinguir los 3 años
colores = {2023: "#2196F3", 2024: "#FF9800", 2025: "#4CAF50"}

# Le pongo un título al gráfico general
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("Partes b) y c) — FDA Empírica y Densidad por Año", fontsize=14, fontweight="bold")

# Grafico cada uno de los años con un color distinto
for año, color in colores.items():
    plot_fda_densidad(df_francisco, "Don Francisco", axes[0, 0], axes[0, 1], año, color)
    plot_fda_densidad(df_miguel,    "Don Miguel",    axes[1, 0], axes[1, 1], año, color)
```

#### Ejercicio 4.d

```python
# Unimos por fecha para tener pares (X_francisco, Y_miguel)
df_conjunto = df_francisco[["fecha", "clientes"]].merge(
    df_miguel[["fecha", "clientes"]], on="fecha", suffixes=("_francisco", "_miguel")
)

# Me quedo con los clientes de cada uno de los almacenes
x = df_conjunto["clientes_francisco"].values
y = df_conjunto["clientes_miguel"].values
 
# Genero el KDE conjunta 
kde2d = gaussian_kde(np.vstack([x, y]), bw_method="scott")

# Defino las grillas [x, y] con las mismas consideraciones de antes
xi = np.linspace(x.min() - 5, x.max() + 5, 100)
yi = np.linspace(y.min() - 5, y.max() + 5, 100)
# Defino una grilla de puntos en 2D
Xi, Yi = np.meshgrid(xi, yi)
# Me quedo con los puntos definidos por la grilla
Zi = kde2d(np.vstack([Xi.ravel(), Yi.ravel()])).reshape(Xi.shape)
 
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Parte d) — Densidad Empírica Conjunta (2023–2025)", fontsize=14, fontweight="bold")
 
# Mapa de calor
im = axes[0].contourf(Xi, Yi, Zi, levels=20, cmap="viridis")
axes[0].set_xlabel("Clientes Don Francisco")
axes[0].set_ylabel("Clientes Don Miguel")
axes[0].set_title("Mapa de densidad conjunta")
plt.colorbar(im, ax=axes[0], label="Densidad")
 
# Scatter con transparencia
axes[1].scatter(x, y, alpha=0.15, s=5, color="#2196F3")
axes[1].set_xlabel("Clientes Don Francisco")
axes[1].set_ylabel("Clientes Don Miguel")
axes[1].set_title("Dispersión de clientes diarios")
axes[1].grid(True, alpha=0.3)
```

#### Mapa de densidad conjunta

Los ejes representan los clientes diarios de cada comerciante
 - En el eje X se ven los clientes de Don Francisco
 - En el eje y los clientes de Don Miguel

Cada punto del plano tiene un color según que tan probable sea esa combinación de clientes.
El centro amarrillo del gráfico representa la combinación de clientes más típica de ambos negocios. Las zonas oscuras en los bordes son combinaciones que casi nunca ocurren.

#### Dispersión de clientes diarios

Cada punto es un día, y su posición indica cuántos clientes tuvo cada comerciante ese día:
 - La nube tiene forma "elíptica" y orientada diagonalmente, lo que indica una correlación positiva entre ambos negocios. Tiene sentido porque comparten los efectos anuales, mensuales y diarios.
 - El ancho de la nube muestra la variabilidad. Si fuera una línea perfecta, los clientes de uno determinarían exactamente los del otro. Como es una nube, hay variación propia de cada negocio.
