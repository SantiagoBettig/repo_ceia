```python
import pandas as pd
import matplotlib as mp

# Set options to display all rows and columns
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
```

```python
smn_df = pd.read_csv('smn_historico.csv')
smn_df.head()
```

```python
# Lo primero que tengo que hacer es quedarme con los valores de la región que elegí
# En mi caso voy a quedarme con NOA: JUJUY, SALTA, TUCUMÁN, CATAMARCA, SANTIAGO DEL ESTERO
smn_df.provincia.unique()
```

```python
# Por ende, tenemos que tomar: JUJUY, SANTIAGO DEL ES, CATAMARCA, SALTA y TUCUMAN
# Los NaN se que corresponden a Coronel Pringles (Buenos Aires) y El Trebol (Santa Fe)

# Creo una lista con todas estas variables
provincias = ['JUJUY', 'SALTA', 'TUCUMAN', 'CATAMARCA', 'SANTIAGO DEL ES']
# Voy a buscar estas variables dentro de la columna "Provincia" y luego voy a conformar otro 
# dataframe con ellas
df_noa = smn_df[smn_df['provincia'].isin(provincias)]
df_noa.head()
```

```python
# Para saber cuantas estaciones meteorológicas tengo en la región tengo que hacer:
df_noa.estacion.unique()
# El total de estaciones sera:
print('El total de estaciones es: ', df_noa.estacion.nunique())
```

```python
# La altura media de las estaciones en la región será sera:
print('La altura media de las estaciones en la región será: ', df_noa.altura.mean())
```

```python
# Alguien tuvo la gran idea de almacenar la temperatura como string, por lo que primero debemos convertir el
# formato a numérico
df_noa['temp'] = pd.to_numeric(df_noa['temp'], errors='coerce')
# La temperatura media de las estaciones en la región será sera:
print('La temperatura media en la región será: ', df_noa.temp.mean())
# Mientras que el desvio estandar es:
print('El desvio estandar de la temperatura en la región será: ', df_noa.temp.std())
```

```python
# Alguien tuvo la gran idea de almacenar la precipitación media como string, por lo que primero debemos convertir el
# formato a numérico
df_noa['prec_mm'] = pd.to_numeric(df_noa['prec_mm'], errors='coerce')
# La temperatura media de las estaciones en la región será sera:
print('La precipitación media en la región será: ', df_noa.prec_mm.mean())
```

```python
idx_temp = df_noa['temp'].idxmax()
idx_prec = df_noa['prec_mm'].idxmax()

print("Mayor temp:")
print(df_noa.loc[idx_temp, ['mes_txt', 'provincia', 'temp']])
print("\nMayor prec_mm:")
print(df_noa.loc[idx_prec, ['mes_txt', 'provincia', 'prec_mm']])
```

```python
def detectar_outliers(df, columna):
    Q1 = df[columna].quantile(0.25)
    Q3 = df[columna].quantile(0.75)
    IQR = Q3 - Q1
    limite_inf = Q1 - 1.5 * IQR
    limite_sup = Q3 + 1.5 * IQR
    outliers = df[(df[columna] < limite_inf) | (df[columna] > limite_sup)]
    print(f"\n--- {columna} ---")
    print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
    print(f"Límite inferior: {limite_inf}, Límite superior: {limite_sup}")
    print(f"Outliers encontrados: {len(outliers)}")
    print(outliers[['estacion', 'provincia', 'mes_txt', columna]])

detectar_outliers(df_noa, 'temp')
detectar_outliers(df_noa, 'prec_mm')
```
